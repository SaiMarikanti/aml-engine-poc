```python
from pyspark.sql.types import (
    StructType,
    StructField,
    LongType,
    StringType,
    DoubleType,
    TimestampType,
    BooleanType
)

transaction_schema = StructType([
    StructField("TX_ID", LongType(), True),
    StructField("SENDER_ACCOUNT_ID", LongType(), True),
    StructField("RECEIVER_ACCOUNT_ID", LongType(), True),
    StructField("TX_TYPE", StringType(), True),
    StructField("TX_AMOUNT", DoubleType(), True),
    StructField("TIMESTAMP", TimestampType(), True),
    StructField("IS_FRAUD", BooleanType(), True),
    StructField("ALERT_ID", LongType(), True)
])
```

```python
raw_tx_stream = (
    spark.readStream
    .format("cloudFiles")
    .option("cloudFiles.format", "csv")
    .option("header", "true")
    .option("cloudFiles.schemaLocation", f"{CHECKPOINT_PATH}/schema_tx")
    .schema(transaction_schema)
    .option("rescuedDataColumn", "_rescued_data")
    .load(f"{S3_RAW_PATH}/transactions/")
)
```
