Rule Engine and later GraphFrames will treat account IDs as graph vertices. If transactions reference nonexistent accounts, your AML network is structurally inconsistent.

```python
%sql
-- Transactions → Accounts
SELECT COUNT(*) AS invalid_sender_accounts
FROM aml_engine.aml_poc.silver_transactions t
LEFT JOIN aml_engine.aml_poc.silver_accounts a
    ON t.sender_account_id = a.account_id
WHERE a.account_id IS NULL;
```

```python
%sql
-- Transactions → Accounts
SELECT COUNT(*) AS invalid_receiver_accounts
FROM aml_engine.aml_poc.silver_transactions t
LEFT JOIN aml_engine.aml_poc.silver_accounts a
    ON t.receiver_account_id = a.account_id
WHERE a.account_id IS NULL;
```

```python
%sql
-- Alerts → Transactions
SELECT COUNT(*) AS orphan_alerts
FROM aml_engine.aml_poc.silver_alerts a
LEFT JOIN aml_engine.aml_poc.silver_transactions t
    ON a.tx_id = t.tx_id
WHERE t.tx_id IS NULL;
```
